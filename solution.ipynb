# Improved constructive solution

This notebook generates `submission_improved.csv` using mathematical constructions, selected published layouts, and outer-row extensions. Proof-backed cases stay fixed. The local validator checks the output and calculates the Kaggle score.

```python
from pathlib import Path
import sys

project_dir = Path.cwd()
if not (project_dir / 'packing_solution.py').exists():
    project_dir = project_dir / 'square_in_squares'
if not (project_dir / 'packing_solution.py').exists():
    raise FileNotFoundError('Run the notebook from square_in_squares or its parent directory')
sys.path.insert(0, str(project_dir))

from packing_solution import write_submission
from area_gap_report import ranked_gaps, write_report
from optimal_locks import check_locked_sides
from validate_submission import total_score, validate_submission

output = project_dir / 'submission_improved.csv'
improved_count = write_submission(output)
results = validate_submission(output)
check_locked_sides(results)
gap_report = project_dir / 'area_gap_report.csv'
write_report(gap_report, ranked_gaps(results))
print(f'Improved puzzles: {improved_count}')
print(f'Kaggle score: {total_score(results):.12f}')
print(f'Submission: {output}')
print(f'Area gap report: {gap_report}')
```
