```python
import csv
import math

with open("submission.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["id", "n", "square_id", "x", "y", "angle"])
    row_id = 0
    for n in range(1, 1001):
        k = math.ceil(math.sqrt(n))
        for i in range(n):
            x = (i % k) + 0.5
            y = (i // k) + 0.5
            writer.writerow([row_id, n, i, x, y, 0.0])
            row_id += 1
print("Total rows:", row_id)
```
